```python
curren_ex = 'Historical Maturity Terms'
sheet = workbook[curren_ex]
cell = sheet.cell(row=17, column=12)
cell.value
cell.number_format
```

```python
sheet_level_fmt = {}
for c_sheet in workbook.sheetnames:
    
    number_formats = []

    sheet = workbook[c_sheet]
    max_row = sheet.max_row
    max_column = sheet.max_column
    for row in range(1, max_row+1):
        for column in range(1, max_column+1):
            cell_obj = sheet.cell(row=row, column=column)
            n_format = cell_obj.number_format
            if n_format not in number_formats:
                number_formats.append(n_format)
    sheet_level_fmt[c_sheet]=number_formats

#             if cell_value:
#                 print(sheet.cell(row=row, column=column).internal_value.formatted_value)
#                 print(cell_value)
        # Perform analysis or checks on the cell value
        # to determine if it belongs to a table
        # For example, you could check for specific patterns or keywords
        
print(sheet_level_fmt)
```
